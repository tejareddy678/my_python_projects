```python
import numpy as np
```

```python
scores = np.array([
    [85, 90, 88],  # Student 1
    [92, 87, 91],  # Student 2
    [78, 82, 80],  # Student 3
    [95, 98, 96],  # Student 4
    [88, 85, 90],  # Student 5
    [100,100,100]  # Student 6
])
```

```python
print("---------------------------Student Statistics--------------------------\n")
average_mark = np.mean(scores,axis = 1)
print("Average mark of each student in each test : ",average_mark)
print("Student with better average mark is : ",np.max(average_mark))
print("Student with better average mark (Index) : ",np.argmax(average_mark))
print("Students with average greater than 90 (Index): ",*np.where(average_mark >= 90))
```

```python
print("---------------------------Test Statistics--------------------------\n")
test_average = np.mean(scores,axis = 0)
print("Each test average : ",test_average)
print(f"Test with highest average : test-{str(np.argmax(test_average)+1)} average = ({np.max(test_average)})")
print(f"Test with lowest average : test-{str(np.argmin(test_average)+1)} average = ({np.min(test_average)})")
# Find students who scored above 90 in all tests
print("Students whose marks greater than 90 in all tests : ")
index = list(np.where(np.all(scores >= 90,axis = 1)))
for i in range(0,len(index[0])):
    print(f"student-{index[0][i]+1}  score({scores[index[0][i]]})")
```
